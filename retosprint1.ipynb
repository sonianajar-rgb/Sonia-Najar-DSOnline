
altas_limpias = []
descartes = []
vistos = []
nombres_vistos = []

for linea in ALTAS_TEXTO.split("\n"):
    # Saltar líneas en blanco sin descartarlas
    if not linea.strip():
        continue

    partes = linea.split(";")

    # Limpiar primero los datos
    nombre = " ".join(partes[0].split()).title()
    email = "".join(partes[1].split()).lower()
    programa = " ".join(partes[2].split())

    # Comprobar el nombre antes que el correo
    if nombre.lower() in nombres_vistos:
        descartes.append([linea, "duplicado"])
        continue

    # Comprobar si el correo es válido
    if "@" not in email or not email.endswith("@bridge.example"):
        descartes.append([linea, "email_invalido"])
        continue

    # Comprobar si el correo ya se ha usado
    if email in vistos:
        descartes.append([linea, "duplicado"])
        continue

    # Guardar los datos válidos
    nombres_vistos.append(nombre.lower())
    vistos.append(email)
    altas_limpias.append([nombre, email, programa])
